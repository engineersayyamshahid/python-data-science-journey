# Basic Numpy
Definition: NumPy (Numerical Python) is a Python library used to perform mathematical calculations and work with arrays of numbers easily.

# 1. Creating Array
An array is a collection of elements of the same data type. NumPy arrays are used to store and perform calculations on multiple values efficiently.

```python
import numpy as np

numbers = np.array([10,20,30,40,50])
print(numbers[1])
print(numbers)
```

# 2. Broadcasting
Broadcasting allows NumPy to perform calculations between arrays of different shapes or between an array and a single number.

```python
import numpy as np
numbers = np.array([5,10,15,20,25])
print(numbers + 5) #Here, NumPy adds 5 to every element automatically.
```

# 3. Vectorization
Vectorization means performing an operation on an entire array at once instead of using a loop to process each element individually.

```python
import numpy as np
numbers = np.array([1, 2, 3, 4])
squares = numbers ** 2  #Here, NumPy calculates the square of every element without using a loop.
print(squares)
```
